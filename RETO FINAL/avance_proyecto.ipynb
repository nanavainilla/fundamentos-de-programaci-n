- Cálculo de interes:

```python
 #puse que la preferencia del usuario vale el 60% y el tiempo consumido vale el 40%

#entonces:
preferencia = 0.60
tiempo_consumido = 0.40

puntaje_final = (afinidad * preferencia) + (minutos_vistos * tiempo_consumido)
```
