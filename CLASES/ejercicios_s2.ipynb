EJERCICIO 5
Crea un programa que pida el tipo de visitante (adulto_mayor, profesor, estudiante, y otro)y calcule el precio de un boleto base de $45 aplicando solo un descuento: adulto mayor 12%, profesor 10%, estudiante 10%, otro 0%

```python
precio = 45
tipo = input("ingrese su tipo de cliente: ")
if tipo == "adulto_mayor":
    descuento = precio * 0.12
elif tipo == "profesor" or tipo == "estudiante": 
    descuento = precio * 0.10
else : descuento = 0

total = precio - descuento
print (f"total a pagar: ${total:.2f}") 
```

EJERCICIO 11 Crea un programa que pida un numero entero positivo N y calcule con "while" la suma de los numeros del 1 al N 

```python
numero = int(input("ingrese un numero: "))
contador = 1
suma = 0

while contador <= numero:
    suma = suma + contador #acumula el valor actual de contador
    contador = contador +1 #el +1 incrementa el contador 1 en 1

print(f"la suma del 1 al {numero} es: {suma}")
```

EJERCICIO 18
Crea un programa que sume numeros capturados con "input"; si el acumulado alcanza o supera 100, el ciclo se detiene con "break" y se muestra el total

```python
while True: 
    numero1 = int(input ("ingrese un numero: "))
    numero2 = int(input ("ingrese un numero: "))
    numero3 = int(input ("ingrese un numero: "))

    suma = 0
    suma = numero3 + numero1 + numero2

    if suma >= 100:
            print ("total: ", suma)
            break
    
```
