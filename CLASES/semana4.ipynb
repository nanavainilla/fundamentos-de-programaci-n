```python
filas = int (input ("ingrese el numero de filas que se desea: "))
columnas = 3

matriz = [] 
for f in range (filas):
    fila_actual = []
    for c in range (columnas): 
        valor = int (input(f"ingresa el valor para [{f}][{c}]: "))
        fila_actual.append(valor)
    matriz.append(fila_actual)

print ("matriiiiiizzzz")

for f in range (filas):
    for c in range (columnas):
        print (matriz[f][c], end = "    ")
    print ("")
```
