```python
def suma_tupla (t) :
    total = 0
    for e in t:
        total += e
    return total 

calorias = (800, 950, 700, 1300, 800, 900, 650)
resultado = suma_tupla(calorias)

print (f"comiste {resultado} calorias, gorda panzona!!")
```

```python
def buscar_contacto (contactos, nombre):
    return contactos.get (nombre, "contacto no encontrado:")

nombre = input ("escribe el nombre del contacto que buscas: ")
telefono = buscar_contacto (contactos, nombre)

print (f"el telefono de {nombre} es: {telefono} ")
```

EJERCICIO EXTRA AUTOEVAUABLE 2 CONTADOR DE VOTOS

```python
candidatos = {
    "A" : 0,
    "B" : 0, 
    "C" : 0  
}

cantidad = int(input ("cuantos votos van a ser?"))

for c in range (1, cantidad + 1):
    voto = input (f"ingresa el voto {c}")
    print (f"votaste por {voto.upper ()}")
    if voto.upper () in candidatos: 
        print (f"el voto esta dentro de los candidatos")
        candidatos [voto.upper ()] += 1
    else: 
        print ("el voto no es valido")



for k, v in candidatos.items (): 
    print (f"opcion {k}, votos: {v}")
```
