EJERCICIO 2
Crea una función `suma_fila(fila)` que **regrese valor** (con
`return`) la suma de los elementos de una lista, y una función
`mostrar_tabla(matriz)` que **no regrese valor** e imprima la matriz sin corchetes
ni comas. Dada la matriz `[[3, 1, 4], [1, 5, 9], [2, 6, 5]]`, imprime la matriz y
la suma de cada fila.

```python
def suma_fila(fila):
    suma = 0
    for elemento in fila:
        suma += elemento
    return(suma)


def mostrar_tabla(matriz):
    for fila in matriz:
        print("")
        for elemento in fila:
            print((elemento), end = "  ")

matriz = [
    [3, 1, 4],
    [1, 5, 9],
    [2, 6, 5]
]

mostrar_tabla(matriz)
suma = 0

print()
for fila in matriz:
    resultado = suma_fila(fila)
    suma = resultado
    print ("la suma de la fila es: ", suma)
```

EJERCICIO 3 
Crea la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]` (4
filas y 3 columnas). Define una función `multiplicar_columna(matriz, columna)` que
**regrese valor** (con `return`) el producto de los elementos de una columna dada.
**Restricción:** dentro de la función está prohibido usar el operador `*`; el
producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el
número de columna (`0`, `1` o `2`) y muestra el resultado.

```python
def multiplicar(a, b):
    resultado = 0
    for i in range(b):
        resultado += a
    return resultado


def columnamult(matriz, columna):
    producto = matriz[0][columna]

    for fila in range(1, len(matriz)):
        numero = matriz[fila][columna]
        producto = multiplicar(producto, numero)

    return producto


matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
    [2, 4, 6]
]

columna = int(input("ingrese el número de columna que desea multiplicar: "))

resultado = columnamult(matriz, columna) 
print("el producto de la columna elegida es:", resultado)
```
