EJERCICIO 6 
Sin usar los operadores `/` ni `//`, calcula el cociente y el
residuo de una división entre enteros positivos restando el divisor al dividendo
con un `while`; el cociente es la cantidad de restas realizadas.

```python
numero1 = int(input("ingrese el dividendo: "))
numero2 = int(input("ingrese el divisor: "))

dividendo = numero1
divisor = numero2
dividendo_original = dividendo
contador = 0

while dividendo >= divisor:
    dividendo -= divisor
    contador += 1
    residuo = dividendo
    print(f"{dividendo_original} + {divisor} = {contador} con residuo {residuo}")
```

EJERCICIO 11 
Crea un programa que pida un entero positivo `n` y calcule la suma
alternada `1 - 2 + 3 - 4 + ...` hasta `n`. Usa un `for` y decide el signo de cada
término con el módulo `%

```python
numero = int(input("ingrese un numero entero positivo: "))
suma = 0

for i in range(1, numero + 1):
    if i % 2 != 0:
        suma += i
    else:
        suma -= i

print(f"el resultado de la suma es: {suma}")
```

EJERCICIO 12
Crea un programa que pida un entero positivo `n` y calcule la suma
`1 + 1/2 + 1/3 + ... + 1/n` con un acumulador flotante. Muestra el resultado con 2

```python
numero = int(input ("ingrese un numero entero positivo: "))
suma = 0 

for i in range(1, numero + 1):
    suma += 1 / i

print(f"la suma de la serie es: {suma:.2f}")
```

EJERCICIO 13 
Crea un programa que pida cuántos términos de la sucesión de
Fibonacci mostrar. La sucesión comienza con `0` y `1`, y cada término siguiente es
la suma de los dos anteriores. Usa dos variables que avancen junto con el `for`

```python
terminos= int(input("cuantos terminos de Fibonacci mostrar?: "))
a, b = 0, 1

fibonacci = []


for _ in range(terminos):
    fibonacci.append(str(a))
    a, b = b, a + b


print("Fibonacci:", ", ".join(fibonacci))
```
