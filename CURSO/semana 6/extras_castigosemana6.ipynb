EJERCICIO 10.

EJERCICIO 12. 
Suma de pares del 1 al 20 con `for` y `continue`

```python
suma = 0

for numero in range(1, 21):
    if numero % 2 != 0:
        continue 
    suma += numero

print("la suma de los numeros pares del 1 al 20 es: ", suma)
```

EJERCICIO 13. 
Suma de una lista con función que regresa valor (sin `sum()`)

```python
def sumar_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total

numeros = [85, 420, 59, 18]

resultado = sumar_lista(numeros)

print("la suma de todos los elementos es: ",resultado)
```

EJERCICIO 14. Máximo de una lista con función que regresa valor (sin `max()`)

```python
def obtener_maximo(lista):
    maximo = lista[0]
    for numero in lista[1:]:
        if numero > maximo:
            maximo = numero
    return maximo

numeros = [48, 74, 129, 95]

resultado = obtener_maximo(numeros)

print("el elemento maximo es: ",resultado)
```

EJERCICIO 15. 
Matriz y suma de la diagonal principal

```python
matriz = [
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9]
    ]

suma_diagonal = 0

for s in range (len(matriz)):
    suma_diagonal += matriz [s][s]

print ("la suma de la diagonal principal es: ",suma_diagonal)
```

EJERCICIO 18. 
Factorial con función recursiva

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial (n - 1)

numero = int(input("ingrese el numero que desee: "))

resultado = factorial(numero)

print(f"el factorial de {numero} es: {resultado}")
```

EJERCICIO 19. 
Tupla: acceso y suma con función

```python
def procesar_tupla(tupla):

    tercer_elemento = tupla[2]
    
    suma_total = 0
    for numero in tupla:
        suma_total += numero
        
    return tercer_elemento, suma_total

numeros = (41, 58, 15, 96, 33, 62)

tercer_elemento, suma = procesar_tupla(numeros)

print(f"el tercer elemento es: {tercer_elemento}")
print(f"la suma de la tupla es: {suma}")
```
