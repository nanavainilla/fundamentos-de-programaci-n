EJERCICIO 17 
Crea un programa que lea tres calificaciones de `calificaciones.txt`, calcule el promedio y lo escriba en el archivo `promedio.txt` con el formato `Promedio: <valor>`.

```python
with open("calificaciones.txt", "w") as archivo:
    archivo.write("8\n9\n10\n")

calificaciones = []
with open("calificaciones.txt", "r") as archivo_in:
    for linea in archivo_in:
        calificaciones.append(float(linea.strip()))

promedio = sum(calificaciones) / len(calificaciones)

with open("promedio.txt", "w") as archivo_out:
    archivo_out.write(f"promedio: {promedio:.1f}")
```

Crea un programa que lea los montos de `ventas.txt` (con `float()`), calcule el total y el promedio, y los escriba en un nuevo archivo `resumen.txt` con dos decimales.

```python
with open("ventas.txt", "w") as archivo:
    archivo.write("100.0\n250.5\n49.5\n")

ventas = []
with open("ventas.txt", "r") as archivo_in:
    for linea in archivo_in:
        ventas.append(float(linea.strip()))

total = sum(ventas)
promedio = total / len(ventas)

with open("resumen.txt", "w") as archivo_out:
    archivo_out.write(f"total de ventas: {total:.2f}\n")
    archivo_out.write(f"promedio por venta: {promedio:.2f}\n")
```

EJERCICIO 22
Crea un programa que lea los números de `numeros2.txt`, separe los pares en `pares.txt` y los impares en `impares.txt`.

```python
with open("numeros2.txt", "w") as archivo:
    archivo.write("1\n2\n3\n4\n5\n")

pares = []
impares = []

with open("numeros2.txt", "r") as archivo:
    for linea in archivo:
        num = int(linea.strip())
        if num % 2 == 0:
            pares.append(num)
        else:
            impares.append(num)

with open("pares.txt", "w") as f_pares:
    for n in pares:
        f_pares.write(f"{n}\n")

with open("impares.txt", "w") as f_impares:
    for n in impares:
        f_impares.write(f"{n}\n")
```

EJERCICIO 27 
Crea un programa que lea las líneas de `datos_malos.txt`, intente convertirlas a `float()` dentro de un `try-except` que capture `ValueError` y sume únicamente las líneas válidas.

```python
with open("datos_malos.txt", "w") as archivo:
    archivo.write("10\nhola\n20\n")

suma = 0.0

with open("datos_malos.txt", "r") as archivo:
    for linea in archivo:
        texto = linea.strip()
        try:
            numero = float(texto)
            suma += numero
        except ValueError:
            print(f"eerrrrroorrrrr: la linea '{texto}' no es un numero valido.")

print(f"la suma de los numeros validos es: {suma}")
```

EJERCICIO 29 
Crea una función `leer_archivo(nombre)` que use `try-except`: si el archivo existe, regresa la cantidad de líneas que contiene; si no existe, regresa el mensaje `Error: no se pudo leer '<nombre>'`. Prueba con `poema.txt` y con `nada.txt`.

```python
with open("poema.txt", "w") as archivo:
    archivo.write("linea 1\linea 2\linea 3\n")

def leer_archivo(nombre):
    try:
        with open(nombre, "r") as archivo:
            lineas = archivo.readlines()
            return f"{nombre} -> {len(lineas)} lineas leidas"
    except FileNotFoundError:
        return f"{nombre} -> errrorr: no se pudo leer '{nombre}'"

print(leer_archivo("poema.txt"))
print(leer_archivo("nada.txt"))
```

EJERCICIO 30 
Crea un programa con un menú de 2 opciones (`1. Leer archivo`, `2. Salir`) controlado por un ciclo. Al leer, pide el nombre del archivo dentro de un `try-except` que capture `FileNotFoundError`. El usuario lee `menu.txt` (existente), luego `inexistente.txt` y finalmente sale.

```python
with open("menu.txt", "w") as archivo:
    archivo.write("hola, mundo")

while True:
    print("1. leer archivo")
    print("2. salir")
    opcion = input("selecciona una opcion: ")
    
    if opcion == "1":
        nombre = input("ingrese el nombre del archivo: ")
        try:
            with open(nombre, "r") as archivo:
                contenido = archivo.read().strip()
                print(f"contenido de {nombre}: {contenido}")
        except FileNotFoundError:
            print(f"errrorrr: no se encontro el archivo '{nombre}'.")
    elif opcion == "2":
        print("hasta luego.")
        break
    else:
        print("opcion invalida.")
```
