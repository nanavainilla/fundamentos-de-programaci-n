Desarrollar un programa en Python para cobrar las entradas de los visitantes
que desean recorrer el Museo de Antropología e Historia, calculando el precio
adecuado para cada visitante y aplicando descuentos por tipo de visitante bajo
estrictas condiciones lógicas. 

El programa debe procesar a los visitantes mediante
un ciclo controlado, aplicar los descuentos con una tabla de verdad que
garantice un único descuento por boleto y desplegar el total detallado de todas
las personas ingresadas.

tipo de visitante 
    - niños menores de 3 años /gratis/ 
    - menores de edad (de 3 a 17 años) /$30/
    - mayores de 18 años /$45/

tabla de descuentos 
    - adulto mayor /12%/
    - profesor / 10%/
    -  estudiante /10%/ 

```python
total = 0
visitantes = int(input("agrega el total de visitantes: "))


for i in range (1,visitantes+1):
    print(f"\n visitante {i}")

    entrada_edad = input(f"ingrese la edad del visitante {i}: ")
    edad = int(entrada_edad)

    if edad <= 3:
        costo = 0

    elif edad <= 18:
        costo = 30

    else:
        costo = 45
        
    if costo == 0:
        print("entrada gratis")
        precioporpersona = 0

    else:
        tipo = input("ingrese el tipo de visitante: ")
        
        if tipo in ["estudiante"]:
            descuento =  0.10
        if tipo in ["profesor"]:
            descuento = 0.10
        elif tipo == "adulto_mayor":
            descuento = costo * 0.12
            
        precio_por_persona = costo - descuento
        
    total += precio_por_persona
    print(f"costo por persona: ${precio_por_persona:.2f}")
        
print("\n" + "=" * 35)
print(f"total final: ${total:.2f}")
print("=" * 35)

#intendo de tabla 
#no me quedó :(
for i in range (1,visitantes+1):
    print(f"\n visitante {i}")

print("persona\precio\descuento")
print(f"visitante1\t{precio_por_persona: .2f}\t{descuento}")


    
```
