Elabora un programa que muestre en pantalla una tabla de Pitágoras 


```python
#encabezado

tabla = []

for i in range(1, 11):
    fila = []
    for j in range(1, 11):
        fila.append(i * j)
    tabla.append(fila)

def genera_encabezado():

    print("    ", end="") 
    for c in range(1, 11): #c = columna
        print(f"{c:4d}", end="") # 4d por 4 espacios entre cada número
    
    print()
    
    for f in range(10):  #f para fila
        print(f"{f + 1:3d}|", end="")
        
        for c in range(10):
            print(f"{tabla[f][c]:4d}", end="") 
        
        print()


def multiplicar(matriz, factor1, factor2):
    return matriz[factor1 - 1][factor2 - 1]

genera_encabezado()

factor1 = int(input("\ningrese el primer factor: "))
factor2 = int(input("ingrese el segundo factor: "))

resultado = multiplicar(tabla, factor1, factor2)

print(f"\nel resultado de la multiplicación es: {resultado}")
```
