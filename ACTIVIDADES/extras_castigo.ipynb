EXTRA 2
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con `items()` el color ganador junto con su total.



```python
votos = {
    "rojo": 0,
    "azul": 0 ,
    "verde": 0,
}

for v in range (5): 
    color = input ("ingrese el color por el que desee votar: ")
    if color in votos:
        votos [color] += 1
    else: 
        print ("el color es invalido!! por favor intente de nuevo")

ganador = ()
maximo_votos = -1

for color, total in votos.items():
    if total > maximo_votos:
        maximo_votos = total 
        ganador = color 

print (f"el ganador es {ganador}!! y cuenta con un total de {maximo_votos} votos!!")
```

EXTRA 3
Escribe un programa que pida al usuario dos números enteros dentro de un bloque `try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo `while`. Luego pide la división del primero entre el segundo, con un bloque específico para `ZeroDivisionError`.



```python
while True:
    try:
        num1 = int(input("ingrese el primer numero: "))
        num2 = int(input("ingrese el segundo numero: "))

        print (num1)
        print (num2)

        resultado = num1 / num2

        print("el resultado de la division de ambos numeros es: ", resultado)
        break

    except ValueError:
        print("ERROR!!! ingrese un numero que SI sea entero\n")

    except ZeroDivisionError:
        print("ERROR!!! no se puede dividir entre cero, intente otra vez\n")
```

EXTRA 4
Crea un programa que reciba una frase, imprima su longitud con `len()`, la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.



```python
def contar_palabras(texto):
    return len(texto.split())

frase = input("ingrese la frase que desee, agrega la palabra `python`: ")

print(f"la longitud de la frase es: {len(frase)}")

print(f"la frase en mayusculas es: {frase.upper()}")

reemplazo = frase.replace("python", "programación")
print(f"el texto reemplazado es: {reemplazo}")

total = contar_palabras(frase)
print(f"el total de las palabras que contiene la frase es de: {total}")
```
