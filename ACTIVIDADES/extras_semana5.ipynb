EJERCICIO 1 
Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.

```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)      
print("la tercera calificación es:", calificaciones[2])

calif_nueva1= float(input("ingrese una calificacion nueva: "))   
calif_nueva2 = float(input("ingrese una segunda calificación nueva: "))

tupla = calificaciones + (calif_nueva1, calif_nueva2)     
print("nueva tupla:", tupla)               

lista = list(tupla)            

lista.sort(reverse=True)               

print("lista de mayor a menor:", lista)     

def suma_calificaciones(lista):                     
    return sum(lista)

total = suma_calificaciones(lista)    
print(f"la suma total de las calificaciones es: {total} ")
```

EJERCICIO 2
Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto.

```python
agenda = {                                 
    "ana": "555-0101",
    "luis": "555-0102",
    "mia": "555-0103"
}

contacto = input("ingrese el contacto del nuevo contacto: ") 
numero_contacto = input("ingrese el numero del nuevo contacto: ")  

agenda[contacto] = numero_contacto
nombres = agenda.keys()                      

print("tus contactos registrados son:", ", ".join(nombres))

def buscar_numero(nombre):             
    return agenda.get(nombre)          

buscar_nombre = input("ingrese el nombre que desees buscar: ")         
telefono = buscar_numero(buscar_nombre)
print(f"El teléfono de {buscar_nombre} es: {telefono}")   
```

EJERCICIO 3 
Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.

```python
try:                              
    num1 = int(input("ingrese el primer numero: ")) 
    num2 = int(input("ingrese el segundo numero: "))

    resultado = num1 / num2  

    print(f"{num1} dividido entre {num2} es: {resultado}")

except ValueError:                 
    print("ERROR!!!! ingrese solo numeros enteros")

except ZeroDivisionError:        
    print("ERROR!!!! no se puede dividir entre cero, ingresa otro numero")
```

EJERCICIO 4
Crea una variable `mensaje = "Python es un lenguaje poderoso"`.
Imprime su longitud con `len()`, conviértelo a mayúsculas, reemplaza la palabra
`"Python"` por `"programación"` y, con una función que retorne el conteo, muestra
cuántas palabras contiene el mensaje original.

```python
mensaje = "python es un lenguaje poderoso"     

print("la longitud del mensaje es:", len(mensaje))  

mayusculas = mensaje.upper()           
print("el mensaje en mayusculas es: ", mayusculas)

reemplazo = mensaje.replace("python", "programación")  
print("el texto reemplazado es:", reemplazo)  

def contar_palabras(texto):      
    return len(texto.split())

total_palabras = contar_palabras(mensaje)         
print("las palabras totales son:", total_palabras)
```
